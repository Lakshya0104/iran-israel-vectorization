```python
!pip install transformers torch

import pandas as pd
import torch
from transformers import AutoTokenizer, AutoModel
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

df = pd.read_csv("iran_israel_tweets.csv")

texts = df["text"].tolist()
labels = df["label"].values

from sklearn.model_selection import train_test_split
X_tr, X_te, y_tr, y_te = train_test_split(texts, labels, test_size=0.2)

model_name = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)

def embed(texts):
    vecs = []
    for t in texts:
        inputs = tokenizer(t, return_tensors="pt", truncation=True, padding=True)
        with torch.no_grad():
            output = model(**inputs)
        vecs.append(output.last_hidden_state[:,0,:].numpy()[0])
    return vecs

X_tr_emb = embed(X_tr)
X_te_emb = embed(X_te)

clf = LogisticRegression(max_iter=1000)
clf.fit(X_tr_emb, y_tr)

pred = clf.predict(X_te_emb)
print("BERT Accuracy:", accuracy_score(y_te, pred))
```
